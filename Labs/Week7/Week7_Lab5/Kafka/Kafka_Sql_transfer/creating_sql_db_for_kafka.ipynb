```python
import sqlite3

# 1. Connect to 'company.db' (SQLite creates this file if it doesn't exist)
conn = sqlite3.connect('mydatabase.db')
cursor = conn.cursor()

# 2. Create a 'employees' table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        department TEXT NOT NULL,
        salary REAL NOT NULL
    )
''')

# 3. Clear old data (optional, so you can re-run safely)
cursor.execute("DELETE FROM employees")

# 4. Insert sample rows
employees_data = [
    ('Alice Smith', 'Engineering', 85000.0),
    ('Bob Jones', 'Marketing', 62000.0),
    ('Charlie Brown', 'Engineering', 91000.0)
]

cursor.executemany(
    "INSERT INTO employees (name, department, salary) VALUES (?, ?, ?)",
    employees_data
)

# 5. Commit changes and close
conn.commit()
conn.close()

print("Database 'company.db' and table 'employees' created successfully!")
```
